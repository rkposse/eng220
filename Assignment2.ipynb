```python
#• Stores the prices of 5 grocery items in a list.
#Asks the user to enter the quantities of each item.
#Calculates and prints the total cost of all items.

print("Welcome to the grocery store!")
print("Here is our selection")
print("Apples, steak, eggs, root beer,Oranges")

eggs=input("How many eggs would you like?")

apples=input("How many apples would you like?")

steak=input("How many steak would you like?")

root_beer=input("How many root beer would you like?")

oranges=input("How many oranges would you like?")

def Groc_Bill(eggs,apples,steak,root_beer,oranges):
    eggs=int(eggs)
    apples=int(apples)
    steak=int(steak)
    root_beer=int(root_beer)
    oranges=int(oranges)
    return 0.60*eggs+2*apples+12*steak+5*root_beer+1*oranges


print(f"Your total today is ${Groc_Bill(eggs,apples,steak,root_beer,oranges)}")
```

```python
#• Takes a sentence as input.
#Splits it into words and stores them in a set.
#Prints the set of unique words and the number of unique words.

input_sentence = input("Enter a sentence: ")
words = input_sentence.split()
unique_words = set(words)
print("Unique words:", unique_words)
print("Number of unique words:", len(unique_words))
```
